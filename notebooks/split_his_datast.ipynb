```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import pickle
```

```python
hisdf = pd.read_csv("/home/labs/fleishman/itayta/new_fitness_repo/fitness_learning/notebooks/data/his/his.csv")
his_cols_all = ["L7", "D211"]
```

```python
segment_2 = [137, 139, 140, 141, 142, 143, 144, 157, 160, 161, 163]
segment_5 = [181, 183, 189, 201, 202, 203, 204, 205, 206, 207, 208, 209, 210, 211]
#segment_7 = [67, 73, 86, 87, 92, 93]


segment_2_wt_aa = "NPYAVVECIPF"
segment_5_wt_aa =  "RKSREATSPNGTND"

segment_2_final_muts = []
segment_5_final_muts = []
#segment_7_wt_aa =  "HSHITT"
```

```python
hisdf.columns

si = np.where(hisdf.columns == his_cols_all[0])[0][0]
ei = np.where(hisdf.columns == his_cols_all[1])[0][0]

his_df_all_muts = hisdf.columns[si:(ei+1)]
all_muts = his_df_all_muts.tolist()
wt_muts = [mp[0] for mp in all_muts]
wt_pos = [int(mp[1:]) for mp in all_muts]

for (wt_aa, seg_pos, fnl_muts) in [(segment_2_wt_aa, segment_2, segment_2_final_muts), (segment_5_wt_aa, segment_5, segment_5_final_muts)]:
    for m,p in zip(wt_aa, seg_pos):
        suggested = "%s%d" % (m,p)

        if suggested not in all_muts:
            print("Got %s in dataset %s" % (suggested, "%s%d" % (wt_muts[wt_pos.index(p)], p)))

        fnl_muts.append(suggested)
```

```python
hisdf_seg_5 = hisdf.copy()#[segment_5_final_muts + ["sanity_mut_numb", "full_seq", "activity", "num_muts"]]
for mt in segment_5_final_muts: 
    print(hisdf_seg_5[mt].unique())


hisdf_5_seg_indices = np.repeat(True, hisdf.shape[0])

for mt in all_muts:

    if mt in segment_5_final_muts: 
        continue

    wt = mt[0]
    hisdf_5_seg_indices = hisdf_5_seg_indices & (hisdf_seg_5[mt] == mt[0])

hisdf_5_ind = np.where(hisdf_5_seg_indices)[0]
hisdf_seg_5_f = hisdf.copy()
hisdf_seg_5_f = hisdf_seg_5_f[segment_5_final_muts + ["sanity_mut_numb", "full_seq", "activity", "num_muts"]]
hisdf_seg_5_f = hisdf_seg_5_f.iloc[hisdf_5_ind]

for mt in segment_5_final_muts: 
    print(hisdf_seg_5_f[mt].unique())

print("Dataset variants %d" % hisdf_seg_5_f.shape[0])
plt.hist(hisdf_seg_5_f["activity"])
```

```python
hisdf_seg_2 = hisdf.copy()
for mt in segment_2_final_muts: 
    print(hisdf_seg_2[mt].unique())


hisdf_2_seg_indices = np.repeat(True, hisdf.shape[0])

for mt in all_muts:

    if mt in segment_2_final_muts: 
        continue

    wt = mt[0]
    hisdf_2_seg_indices = hisdf_2_seg_indices & (hisdf_seg_2[mt] == mt[0])

hisdf_2_ind = np.where(hisdf_2_seg_indices)[0]
hisdf_seg_2_f = hisdf.copy()
hisdf_seg_2_f = hisdf_seg_2_f[segment_2_final_muts + ["sanity_mut_numb", "full_seq", "activity", "num_muts"]]
hisdf_seg_2_f = hisdf_seg_2_f.iloc[hisdf_2_ind]

for mt in segment_2_final_muts: 
    print(hisdf_seg_2_f[mt].unique())

print("Dataset variants %d" % hisdf_seg_2_f.shape[0])
#plt.hist(hisdf_seg_2_f["activity"])
#plt.hist(hisdf_seg_2_f["num_muts"])

hisdf_seg_5_f
```
